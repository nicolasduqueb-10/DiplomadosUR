### 📊 **Definición de Covarianza**

$$ COV(X, Y) = \frac{\sum_{i=1}^{n} (x_i - \mu_X)(y_i - \mu_Y)}{n}$$

* $n := $ Cantidad de datos
* $\mu_X :=$ Media de la variable $X$
* $\mu_Y :=$ Media de la variable $Y$
