# 🔄 **Gradiente Descendiente**

El **Gradiente Descendiente** es un algoritmo de **optimización iterativo** que se utiliza para **minimizar una función de costo** $J(\theta)$.  
La idea es actualizar los parámetros en la dirección **opuesta al gradiente**, es decir, hacia donde la función disminuye.


### **📍 Fórmula general de actualización**


$$
\theta := \theta - \alpha \cdot \nabla_\theta J(\theta)
$$

donde:  
- $\theta$ son los parámetros del modelo.  
- $\alpha$ es la *tasa de aprendizaje* o *learning rate*.  
- $\nabla_\theta J(\theta)$ es el gradiente de la función de costo.  

#### **📍 Puntos clave**
- El **gradiente $\nabla_\theta J(\theta)$** indica la dirección de mayor aumento.  
- Para minimizar, avanzamos en la **dirección contraria** $\rightarrow -\nabla_\theta J(\theta)$
- Repetimos el proceso hasta acercarnos al **mínimo de la función de costo**.

### 🧮 **Ejemplo 1: Gradiente Descendiente en 1D**

Para comprender la mecánica del gradiente descendiente, empezamos con una función muy simple:

$$
J(\theta) = \theta^2
$$

- El mínimo global está en $\theta = 0$.  
- El gradiente es:

$$
\nabla J(\theta) = \frac{d}{d\theta} (\theta^2) = 2\theta
$$

La regla de actualización se convierte en:

$$
\theta := \theta - \alpha \cdot (2\theta)
$$

donde $\alpha$ es la tasa de aprendizaje (*learning rate*).  

📌 Este ejemplo nos permitirá ver cómo, a partir de un valor inicial, $\theta$ se va moviendo paso a paso hacia el mínimo en $0$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def J_theta(theta):
    return theta**2

def grad_J_theta(theta):
    return 2*theta
```

```python
theta = 5.0       # inicialización
alpha = 0.8      # learning rate
n_iters = 20      # iteraciones

theta_hist = [theta]
cost_hist = [J_theta(theta)]
```

```python
for _ in range(n_iters):
    theta = theta - alpha * grad_J_theta(theta)  # actualización GD
    theta_hist.append(theta)
    cost_hist.append(J_theta(theta))
```

```python
print(f"θ inicial = {theta_hist[0]:.4f}")
print(f"θ final   = {theta_hist[-1]:.6f}")
print(f"J(θ) final= {cost_hist[-1]:.6f}")
```

#### **📉 Visualización**

```python
# Visualización: función y trayectoria de GD
thetas_plot = np.linspace(-6, 6, 400)
J_plot = J_theta(thetas_plot)

plt.figure()
plt.plot(thetas_plot, J_plot, label="J(θ) = θ^2")
plt.scatter(theta_hist, cost_hist, s=30, label="Trayectoria GD")
plt.title("GD en J(θ)=θ^2: trayectoria hacia el mínimo")
plt.xlabel("θ")
plt.ylabel("J(θ)")
plt.legend()
plt.show()
```

```python
# Visualización del costo
plt.figure()
plt.plot(range(len(cost_hist)), cost_hist, marker="o")
plt.title("Convergencia de GD: J(θ) por iteración")
plt.xlabel("Iteración")
plt.ylabel("J(θ)")
plt.show()
```

### 📈 **Ejemplo 2: Gradiente Descendiente en Regresión Lineal Simple**



En este ejemplo aplicamos **Gradiente Descendiente** para ajustar una recta de regresión lineal simple a un conjunto de puntos.

El modelo es:

$$
\hat{y} = \beta_0 + \beta_1 x
$$

La función de costo que queremos minimizar es el **Error Cuadrático Medio (MSE):**

$$
J(\beta_0, \beta_1) = \frac{1}{2m}\sum_{i=1}^m \big( \hat{y}_i - y_i \big)^2
$$

Los gradientes respecto a los parámetros son:

$$
\frac{\partial J}{\partial \beta_0} = \frac{1}{m}\sum_{i=1}^m \left(\hat{y}_i - y_i \right)
$$

$$
\frac{\partial J}{\partial \beta_1} = \frac{1}{m}\sum_{i=1}^m \left(\hat{y}_i - y_i\right) x_i
$$

Con estos gradientes, las reglas de actualización son:

$$
\beta_0 := \beta_0 - \alpha \cdot \frac{\partial J}{\partial \beta_0}
$$

$$
\beta_1 := \beta_1 - \alpha \cdot \frac{\partial J}{\partial \beta_1}
$$

📌 Iterando este proceso, los coeficientes $(\beta_0, \beta_1)$ se ajustan progresivamente para minimizar el error y encontrar la recta que mejor se ajusta a los datos.

```python
# Datos (x, y)
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4.5, 6], dtype=float)
m = len(x)
```

```python
# Inicialización de parámetros
b0 = 1.5
b1 = 0.8
alpha = 0.05
n_iters = 50
```

```python
def predict(x, b0, b1):
    return b0 + b1 * x

def mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)
```

```python
mse_hist = []

for it in range(n_iters):
    y_hat = predict(x, b0, b1)
    error = y_hat - y

    # Gradientes (derivadas parciales de MSE)
    db0 = (1/m) * np.sum(error)
    db1 = (1/m) * np.sum(error * x)

    # Actualización GD
    b0 -= alpha * db0
    b1 -= alpha * db1
    mse_hist.append(mse(y, predict(x, b0, b1)))
```

```python
print(f"β0 (intercepto) final = {b0:.6f}")
print(f"β1 (pendiente)  final = {b1:.6f}")
print(f"MSE final              = {mse_hist[-1]:.6f}")
```

#### 📉 **Visualización**

```python
# Visualización: datos y recta ajustada con GD
x_line = np.linspace(x.min(), x.max(), 100)
y_line = predict(x_line, b0, b1)

plt.figure()
plt.scatter(x, y, label="Datos")
plt.plot(x_line, y_line, label="Recta (GD manual)")
plt.title("Regresión Lineal Simple: ajuste con GD manual")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
plt.figure()
plt.plot(range(1, n_iters+1), mse_hist)
plt.title("Convergencia de GD en RL Simple (MSE por iteración)")
plt.xlabel("Iteración")
plt.ylabel("MSE")
plt.show()
```

### 👾 **[EXTRA] Elección del $\alpha$**

En Gradiente Descendiente, la **tasa de aprendizaje** $\alpha$ controla el tamaño del paso que damos en cada iteración al actualizar los parámetros.  
Matemáticamente, la regla de actualización es:

$$
\theta := \theta - \alpha \cdot \nabla_\theta J(\theta)
$$

Elegir un buen valor de $\alpha$ es crítico para garantizar que el algoritmo **converja al mínimo de la función de costo**.



📌 **Conceptos clave (nivel básico):**
- **Gradiente grande:** indica una pendiente pronunciada; un paso demasiado grande puede hacer que el algoritmo *salte* el mínimo.  
- **Gradiente pequeño:** indica que estamos cerca del mínimo; los pasos se hacen naturalmente más cortos.  
- **$\alpha$ muy grande:** provoca divergencia o que el algoritmo oscile sin llegar al mínimo.  
- **$\alpha$ muy pequeño:** asegura convergencia pero con pasos tan diminutos que el proceso es extremadamente lento.

#### 📚 **Conceptos matemáticos más avanzados:**

- **Condición de convergencia:** si la función de costo $J(\theta)$ es convexa y diferenciable, el GD converge al mínimo global si  
  $$
  0 < \alpha < \frac{2}{L}
  $$
  donde $L$ es la **constante de Lipschitz** del gradiente (ligada a la curvatura de $J$).  
- **Hessiano ($H$):** matriz de segundas derivadas que describe la curvatura. El valor de $\alpha$ debe adaptarse al **autovalor máximo** de $H$ para garantizar estabilidad.  
- **Tasa de convergencia:** en funciones cuadráticas, GD converge linealmente con razón  
  $$
  \rho = \frac{\lambda_{\max} - \lambda_{\min}}{\lambda_{\max} + \lambda_{\min}}
  $$
  donde $\lambda_{\max}, \lambda_{\min}$ son los autovalores de $H$.  
- **Métodos adaptativos:** optimizadores modernos (Adam, RMSProp, Adagrad) ajustan $\alpha$ dinámicamente según el comportamiento del gradiente.
